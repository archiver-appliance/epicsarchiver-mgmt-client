```python
path = "AS-243-followup/storagelog/pv_files_bkp-2025-01-28"
import re

def convert_to_filename(pv_name):
    return pv_name.replace(":", "/").replace("-", "/")

def title_pv_name(pv_name):
    return f"### pvNmae: {convert_to_filename(pv_name)}"

def find_linac_number(line):
    pattern = r"archiver-linac-(\d\d)"
    match = re.search(pattern, line)
    if match:
        return match.group(1)
    else:
        return None

def parse_file(path):
    result = {}
    with open(path, 'r') as f:
        pv = None
        for line in f:
            linac_number = find_linac_number(line)
            pv_name = [(from_pv, to_pv) for from_pv, to_pv in failed_pvs if title_pv_name(from_pv) in line or title_pv_name(to_pv) in line]
            if pv_name:
                from_pv, to_pv = pv_name[0]
                if convert_to_filename(from_pv) in line:
                   pv = from_pv
                elif convert_to_filename(to_pv) in line:
                   pv = to_pv
                print(f"pv_name: {pv_name}, pv: {pv}, linac_number: {linac_number}")

            if pv and linac_number:
                
                print(f"pv: {pv}, linac_number: {linac_number}")
                if linac_number:
                    result[pv] = linac_number
                pv = None
    return result

    
```

```python
pvs_linacs = parse_file(path)
```

```python
import csv

updates_pvs_filename = "AS-243-followup/UpdateRequest.csv"

pv_updates_reader = csv.reader(open(updates_pvs_filename))

pv_renames = []
pv_deletes = []
for row in pv_updates_reader:
    old_pv = row[0].strip().split("'")
    if len(old_pv) != 3:
        continue
    col2 = row[1].strip()
    if col2 == "Delete":
        pv_deletes.append(old_pv[1])
    else:
      pv_renames.append((old_pv[1], col2.split("'")[0]))

print(pv_renames)
print(pv_deletes)
```

```python
for from_pv, to_pv in pv_renames:
    from_status = archiver_linac.get_archiving_status(from_pv)
    to_status = archiver_linac.get_archiving_status(to_pv)
    LOG.info(f"from {from_pv} is {from_status} to {to_pv} is {to_status}")
```

```python

def pause_multiple_pvs(archiver, pv_list):
    for pv in pv_list:
        LOG.info(f"Pausing PV: {pv}")
        pause_result = archiver.pause_pv(pv)
        LOG.info(f"Pause result: {pause_result}")

def rename_pv(archiver, old_pv, new_pv) -> dict[str, dict[str, str]]:
    pause_multiple_pvs(archiver, [old_pv, new_pv])
    LOG.info(f"Renaming {old_pv} to {new_pv}")
    new_pv_status = archiver.get_archiving_status(new_pv)
    LOG.info(f"New PV status: {new_pv_status}")
    if new_pv_status == ArchivingStatus.Paused:
        archiver.rename_and_append(old_pv, new_pv, Storage.MTS)
        LOG.info(f"Renamed {old_pv} to {new_pv}")
    new_pv_status = archiver.get_archiving_status(new_pv)
    old_pv_status = archiver.get_archiving_status(old_pv)
    if new_pv_status == ArchivingStatus.Paused:
        LOG.info(f"Resuming {new_pv}")
        archiver.resume_pv(new_pv)
        new_pv_status = archiver.get_archiving_status(new_pv)
        old_pv_status = archiver.get_archiving_status(old_pv)
    return {new_pv: {"status": new_pv_status, "old_pv": old_pv, "old_pv_status": old_pv_status}}

def rename_multiple_pvs(archiver, pv_renames):
    for old_pv, new_pv in pv_renames:
        rename_pv(archiver, old_pv, new_pv)
```

```python
rename_multiple_pvs(archiver_linac, pv_renames)
```
