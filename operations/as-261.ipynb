```python
prefix = "Tgt-HePC1011:Proc-PT-020a"
new_prefix = "Tgt-HePC1011:Proc-PT-017"
pvs = [
":AlmNotConf",
":BSQ",
":FB_AlaHH",
":FB_AlaHHDel",
":FB_AlaHH_D",
":FB_AlaLL",
":FB_AlaLLDel",
":FB_AlaLL_D",
":FB_Ala_D",
":FB_ForceValue",
":FB_ProcHi",
":FB_ProcHiDelay",
":FB_ProcHiHi",
":FB_ProcHiHiDelay",
":FB_ProcLo",
":FB_ProcLoDelay",
":FB_ProcLoLo",
":FB_ProcLoLoDelay",
":FB_WarH",
":FB_WarHDel",
":FB_WarH_D",
":FB_WarL",
":FB_WarLDel",
":FB_WarL_D",
":HiHiSeverity",
":HiSeverity",
":LoLoSeverity",
":LoSeverity",
":MeasCurrent",
":MeasValue",
":MeasVoltage",
":NoProcAlarms",
":P_AlaHH",
":P_AlaHHDel",
":P_AlaLL",
":P_AlaLLDel",
":P_ForceValue",
":P_WarH",
":P_WarHDel",
":P_WarL",
":P_WarLDel",
":ProcHi",
":ProcHiHi",
":ProcLo",
":ProcLoLo",
":RAWValue",

]
rename_pvs = [(prefix + pv, new_prefix + pv) for pv in pvs]
```

```python
from epicsarchiver import ArchiverAppliance
archiver_linac = ArchiverAppliance("archiver-linac-04.tn.esss.lu.se")
```

```python
for from_pv, to_pv in rename_pvs:
    from_status = archiver_linac.get_archiving_status(from_pv)
    to_status = archiver_linac.get_archiving_status(to_pv)
    print(f"from {from_pv} is {from_status} to {to_pv} is {to_status}")
```

```python
for from_pv, to_pv in rename_pvs:
    archiver_linac.pause_pv(from_pv)
    archiver_linac.pause_pv(to_pv)
```

```python
for from_pv, to_pv in rename_pvs:
    archiver_linac.rename_pv(from_pv, to_pv)
```

```python
for from_pv, to_pv in rename_pvs:
    from_status = archiver_linac.get_archiving_status(from_pv)
    to_status = archiver_linac.get_archiving_status(to_pv)
    print(f"from {from_pv} is {from_status} to {to_pv} is {to_status}")
```

```python
for from_pv, to_pv in rename_pvs:
    archiver_linac.resume_pv(to_pv)
```

```python
for from_pv, to_pv in rename_pvs:
    from_status = archiver_linac.get_archiving_status(from_pv)
    to_status = archiver_linac.get_archiving_status(to_pv)
    print(f"from {from_pv} is {from_status} to {to_pv} is {to_status}")
```
